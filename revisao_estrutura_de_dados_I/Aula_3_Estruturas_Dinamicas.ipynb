````python
# Conteúdo de Revisão — Aula 03: Ponteiros, Referências e Estruturas Dinâmicas em Python

## Estrutura de Dados

**Duração:** 1h30
**Linguagem:** Python
**Tema:** Ponteiros de memória, referências e estruturas dinâmicas.

## 1. Objetivos de aprendizagem

Ao final da aula, o estudante deverá ser capaz de:

- compreender o conceito de endereço de memória;
- diferenciar ponteiros de referências;
- utilizar `id()` para observar a identidade de objetos;
- compreender compartilhamento de referências;
- diferenciar referência e cópia;
- compreender objetos mutáveis e imutáveis;
- criar estruturas dinâmicas utilizando referências;
- implementar nós (`Node`) e listas encadeadas;
- utilizar técnicas básicas de depuração;
- trabalhar colaborativamente.

> **Importante:** Python não possui ponteiros explícitos como C/C++. Nesta aula, o conceito de ponteiro é usado para estabelecer a base conceitual para referências e estruturas dinâmicas.

## 2. Memória e variáveis

Considere:

```python
nome = "Ana"
idade = 20
```

Uma forma didática de visualizar:

```text
nome ─────→ "Ana"
idade ────→ 20
```

Em Python, é mais adequado pensar que uma variável é um **nome associado a um objeto**.

## 3. O conceito de ponteiro

Em C:

```c
int idade = 20;
int *p = &idade;
```

Conceitualmente:

```text
idade
  │
  ▼
┌──────────┐
│    20    │
└──────────┘
     ▲
     │
     p
```

- `&` obtém o endereço;
- `*` permite trabalhar com o conteúdo apontado.

Python abstrai esse nível de manipulação.

## 4. Referências em Python

Podemos observar a identidade de um objeto com:

```python
idade = 20
print(idade)
print(id(idade))
```

`id()` retorna um identificador associado à identidade do objeto durante sua existência.

## 5. Compartilhamento de referências

```python
a = [10, 20, 30]
b = a

print(id(a))
print(id(b))
```

`a` e `b` referenciam o mesmo objeto.

```text
        ┌────────────────┐
        │ [10, 20, 30]   │
        └────────────────┘
             ▲       ▲
             │       │
             a       b
```

## 6. Referência versus cópia

### Referência

```python
a = [10, 20, 30]
b = a
b.append(40)

print(a)
print(b)
```

Resultado:

```text
[10, 20, 30, 40]
[10, 20, 30, 40]
```

### Cópia

```python
a = [10, 20, 30]
b = a.copy()
b.append(40)

print(a)
print(b)
```

Resultado:

```text
[10, 20, 30]
[10, 20, 30, 40]
```

## 7. Mutabilidade

Objetos mutáveis incluem:

```text
list
dict
set
```

Exemplo:

```python
lista = [1, 2, 3]
lista.append(4)
print(lista)
```

Objetos como `int`, `float`, `str`, `tuple` e `bool` são imutáveis.

## 8. Passagem de objetos para funções

```python
def adicionar_item(lista):
    lista.append(100)

numeros = [1, 2, 3]
adicionar_item(numeros)

print(numeros)
```

Resultado:

```text
[1, 2, 3, 100]
```

A função recebeu acesso ao mesmo objeto lista.

## 9. Estruturas dinâmicas

Estruturas dinâmicas podem crescer ou diminuir durante a execução.

Exemplos:

- listas;
- filas;
- pilhas;
- listas encadeadas;
- árvores;
- grafos.

Em Python:

```python
dados = []
dados.append(10)
dados.append(20)
dados.append(30)
print(dados)
```

## 10. Nó (`Node`)

```python
class Node:

    def __init__(self, valor):
        self.valor = valor
        self.proximo = None
```

O atributo `proximo` é uma referência para outro nó.

## 11. Criando e conectando nós

```python
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.proximo = node2
node2.proximo = node3
```

Representação:

```text
10 ───→ 20 ───→ 30 ───→ None
```

## 12. Percorrendo os nós

```python
atual = node1

while atual is not None:
    print(atual.valor)
    atual = atual.proximo
```

Saída:

```text
10
20
30
```

## 13. Lista encadeada

```python
class ListaEncadeada:

    def __init__(self):
        self.inicio = None

    def adicionar(self, valor):
        novo = Node(valor)
        novo.proximo = self.inicio
        self.inicio = novo

    def exibir(self):
        atual = self.inicio

        while atual is not None:
            print(atual.valor)
            atual = atual.proximo
```

Uso:

```python
lista = ListaEncadeada()
lista.adicionar(10)
lista.adicionar(20)
lista.adicionar(30)
lista.exibir()
```

## 14. Depuração

Código com erro:

```python
atual = node1

while atual:
    print(atual.valor)
```

O programa entra em loop infinito porque `atual` nunca é atualizado.

Correção:

```python
atual = node1

while atual:
    print(atual.valor)
    atual = atual.proximo
```

Ao depurar, pergunte:

1. Qual variável está incorreta?
2. Qual é o valor dela?
3. Qual objeto ela referencia?
4. Essa referência deveria mudar?
5. Qual condição encerra o `while`?

## 15. Exercício 1 — Referências

Execute:

```python
a = [10, 20, 30]
b = a

b.append(40)

print("a:", a)
print("b:", b)
print("id(a):", id(a))
print("id(b):", id(b))
```

Responda:

1. Qual será a saída?
2. `a` e `b` representam o mesmo objeto?
3. Por quê?
4. O que aconteceria com `b = a.copy()`?

## 16. Exercício 2 — Construindo uma cadeia

Crie:

```python
n1 = Node("A")
n2 = Node("B")
n3 = Node("C")
```

Conecte:

```text
A → B → C → None
```

Depois percorra a estrutura e produza:

```text
A
B
C
```

## 17. Exercício 3 — Depuração

Analise:

```python
n1 = Node(10)
n2 = Node(20)
n3 = Node(30)

n1.proximo = n2
n2.proximo = n3

atual = n1

while atual is not None:
    print(atual.valor)
```

Perguntas:

1. Qual é o problema?
2. Por que o programa não termina?
3. Qual linha deve ser acrescentada?
4. Qual será a saída depois da correção?

## 18. Problema real — Sistema de atendimento de uma clínica

Uma clínica recebe pacientes durante o dia. Cada paciente possui:

- nome;
- idade;
- prioridade.

Exemplo:

```text
Ana — 32 anos — Normal
Bruno — 70 anos — Prioridade
Carlos — 45 anos — Normal
```

O sistema deve permitir:

1. adicionar paciente;
2. listar pacientes;
3. atender o primeiro paciente;
4. verificar se a fila está vazia;
5. informar a quantidade de pacientes.

## 19. Modelagem

Crie:

```python
class Paciente:
    ...
```

Depois:

```python
class Node:
    ...
```

O nó deverá armazenar:

```text
paciente
próximo
```

## 20. Fila de atendimento

Crie:

```python
class FilaAtendimento:
    ...
```

A fila deverá possuir:

```text
início
fim
```

Representação:

```text
início                         fim
  ↓                             ↓
Ana  →  Bruno  →  Carlos  →  None
```

## 21. Funcionalidades obrigatórias

Implemente:

- `adicionar(paciente)`
- `atender()`
- `listar()`
- `esta_vazia()`
- `tamanho()`

## 22. Exemplo de utilização

```python
fila = FilaAtendimento()

fila.adicionar(Paciente("Ana", 32, "Normal"))
fila.adicionar(Paciente("Bruno", 70, "Prioridade"))
fila.adicionar(Paciente("Carlos", 45, "Normal"))

fila.listar()

paciente = fila.atender()
print("Atendendo:", paciente.nome)
```

## 23. Desafio — Atendimento prioritário

Amplie o sistema.

Pacientes com prioridade devem ser atendidos antes dos pacientes normais.

Exemplo:

```text
Fila inicial:

Ana    - Normal
Bruno  - Normal
Carlos - Prioridade
```

Ordem esperada:

```text
1. Carlos
2. Ana
3. Bruno
```

Antes de programar:

1. descreva a solução;
2. desenhe a estrutura;
3. escolha onde o paciente será inserido;
4. implemente;
5. teste.

## 24. Trabalho colaborativo

Forme um grupo de 2 a 4 estudantes.

### Etapa 1 — Modelar

```text
Paciente → Node → Node → Node
```

### Etapa 2 — Implementar

Criar:

```text
Paciente
Node
FilaAtendimento
```

### Etapa 3 — Testar

Teste:

- fila vazia;
- um paciente;
- vários pacientes;
- atendimento;
- remoção do último paciente;
- paciente prioritário.

### Etapa 4 — Depurar

Encontre e corrija pelo menos um erro proposital.

## 25. Exercício de revisão — Sistema de atendimento

Desenvolva individualmente uma versão funcional.

Checklist:

- [ ] Criar `Paciente`;
- [ ] Criar `Node`;
- [ ] Criar `FilaAtendimento`;
- [ ] Implementar `adicionar()`;
- [ ] Implementar `atender()`;
- [ ] Implementar `listar()`;
- [ ] Implementar `esta_vazia()`;
- [ ] Implementar `tamanho()`;
- [ ] Testar pelo menos cinco pacientes;
- [ ] Documentar um erro encontrado e sua correção.

## 26. Checklist de aprendizagem

- [ ] Entendi o conceito de endereço de memória.
- [ ] Sei diferenciar ponteiro de referência.
- [ ] Sei utilizar `id()`.
- [ ] Entendi compartilhamento de referências.
- [ ] Sei diferenciar referência e cópia.
- [ ] Entendi objetos mutáveis.
- [ ] Entendi passagem de objetos para funções.
- [ ] Consigo criar um `Node`.
- [ ] Consigo conectar nós.
- [ ] Consigo percorrer uma estrutura encadeada.
- [ ] Consigo identificar erros em uma estrutura dinâmica.
- [ ] Consigo implementar uma fila.
- [ ] Consigo explicar minha solução.

## 27. Reflexão final

### O que é uma referência em Python?

```text
____________________________________________________
____________________________________________________
```

### Qual a diferença entre `b = a` e `b = a.copy()`?

```text
____________________________________________________
____________________________________________________
```

### Por que uma estrutura encadeada precisa de referências?

```text
____________________________________________________
____________________________________________________
```

### Qual erro você encontrou durante a depuração?

```text
____________________________________________________
____________________________________________________
```

### Como referências ajudam a compreender estruturas dinâmicas?

```text
____________________________________________________
____________________________________________________
```

## 28. Síntese da aula

```text
VARIÁVEL
   ↓
REFERÊNCIA
   ↓
OBJETO
   ↓
NÓ
   ↓
ESTRUTURA DINÂMICA
   ↓
LISTA / FILA / PILHA
```

> **Uma boa estrutura de dados organiza os dados de forma que as operações necessárias sejam realizadas de maneira clara e eficiente.**

### Próxima aula

**Filas e Pilhas**

```text
Lista Encadeada
       ↓
Fila
       ↓
Pilha
       ↓
Árvore
       ↓
Grafo
```

**Boas práticas!**
````

````python
# ============================================================
# AULA 03 - PONTEIROS, REFERÊNCIAS E ESTRUTURAS DINÂMICAS
# ESTRUTURA DE DADOS - PYTHON
# ============================================================


# ============================================================
# 1. REFERÊNCIAS EM PYTHON
# ============================================================

print("=" * 60)
print("EXERCÍCIO 1 - REFERÊNCIAS")
print("=" * 60)

a = [10, 20, 30]
b = a

b.append(40)

print("a:", a)
print("b:", b)
print("id(a):", id(a))
print("id(b):", id(b))
print("a is b:", a is b)

# Respostas:
#
# 1. Qual será a saída?
# a: [10, 20, 30, 40]
# b: [10, 20, 30, 40]
# Os valores de id(a) e id(b) serão iguais, mas o número exato
# depende da execução do programa.
#
# 2. a e b representam o mesmo objeto?
# Sim.
#
# 3. Por quê?
# Porque b = a faz b receber uma referência para o mesmo objeto
# que a referencia. Não é criada uma nova lista.
#
# 4. O que aconteceria com b = a.copy()?
# Seria criada uma nova lista. Alterar b não alteraria a lista a.


# Demonstração da cópia:

print("\n" + "-" * 60)
print("REFERÊNCIA x CÓPIA")
print("-" * 60)

a = [10, 20, 30]

# Referência
b = a
b.append(40)

print("Depois de b = a:")
print("a:", a)
print("b:", b)
print("Mesmo objeto?", a is b)

# Cópia
a = [10, 20, 30]
b = a.copy()
b.append(40)

print("\nDepois de b = a.copy():")
print("a:", a)
print("b:", b)
print("Mesmo objeto?", a is b)


# ============================================================
# 2. MUTABILIDADE
# ============================================================

print("\n" + "=" * 60)
print("MUTABILIDADE")
print("=" * 60)

lista = [1, 2, 3]

print("Lista antes:", lista)

lista.append(4)

print("Lista depois:", lista)

# Listas são objetos mutáveis.
#
# Exemplos de objetos mutáveis:
# list
# dict
# set
#
# Exemplos de objetos imutáveis:
# int
# float
# str
# tuple
# bool


# ============================================================
# 3. PASSAGEM DE OBJETOS PARA FUNÇÕES
# ============================================================

print("\n" + "=" * 60)
print("PASSAGEM DE OBJETOS PARA FUNÇÕES")
print("=" * 60)


def adicionar_item(lista):
    lista.append(100)


numeros = [1, 2, 3]

print("Antes:", numeros)

adicionar_item(numeros)

print("Depois:", numeros)

# Resultado esperado:
# [1, 2, 3, 100]
#
# A função recebeu acesso ao mesmo objeto lista.


# ============================================================
# 4. CLASSE NODE
# ============================================================

print("\n" + "=" * 60)
print("CLASSE NODE")
print("=" * 60)


class Node:
    def __init__(self, valor):
        self.valor = valor
        self.proximo = None


# ============================================================
# 5. CONSTRUINDO UMA CADEIA
# ============================================================

print("\n" + "=" * 60)
print("EXERCÍCIO 2 - CONSTRUINDO UMA CADEIA")
print("=" * 60)

n1 = Node("A")
n2 = Node("B")
n3 = Node("C")

# Conectando os nós
n1.proximo = n2
n2.proximo = n3

# Estrutura:
#
# A -> B -> C -> None

print("Estrutura:")
print("A -> B -> C -> None")

print("\nPercorrendo a estrutura:")

atual = n1

while atual is not None:
    print(atual.valor)
    atual = atual.proximo


# ============================================================
# 6. EXERCÍCIO 3 - DEPURAÇÃO
# ============================================================

print("\n" + "=" * 60)
print("EXERCÍCIO 3 - DEPURAÇÃO")
print("=" * 60)

n1 = Node(10)
n2 = Node(20)
n3 = Node(30)

n1.proximo = n2
n2.proximo = n3

atual = n1

# O código original tinha:
#
# while atual is not None:
#     print(atual.valor)
#
# O problema é que atual nunca era atualizado.
# Isso causaria um loop infinito.
#
# A correção é:
# atual = atual.proximo

print("Estrutura corrigida:")

while atual is not None:
    print(atual.valor)
    atual = atual.proximo

# Respostas:
#
# 1. Qual é o problema?
# A variável atual nunca era atualizada.
#
# 2. Por que o programa não termina?
# Porque atual continua apontando para o mesmo Node.
# Portanto, atual nunca se torna None.
#
# 3. Qual linha deve ser acrescentada?
# atual = atual.proximo
#
# 4. Qual será a saída?
# 10
# 20
# 30


# ============================================================
# 7. LISTA ENCADEADA
# ============================================================

print("\n" + "=" * 60)
print("LISTA ENCADEADA")
print("=" * 60)


class ListaEncadeada:

    def __init__(self):
        self.inicio = None

    def adicionar(self, valor):
        novo = Node(valor)

        novo.proximo = self.inicio

        self.inicio = novo

    def exibir(self):
        atual = self.inicio

        while atual is not None:
            print(atual.valor)
            atual = atual.proximo


lista = ListaEncadeada()

lista.adicionar(10)
lista.adicionar(20)
lista.adicionar(30)

print("Elementos da lista:")
lista.exibir()


# ============================================================
# 8. CLASSE PACIENTE
# ============================================================

print("\n" + "=" * 60)
print("SISTEMA DE ATENDIMENTO DA CLÍNICA")
print("=" * 60)


class Paciente:

    def __init__(self, nome, idade, prioridade):
        self.nome = nome
        self.idade = idade
        self.prioridade = prioridade

    def __str__(self):
        return (
            f"{self.nome} - "
            f"{self.idade} anos - "
            f"{self.prioridade}"
        )


# ============================================================
# 9. NODE PARA A FILA
# ============================================================

class NodePaciente:

    def __init__(self, paciente):
        self.paciente = paciente
        self.proximo = None


# ============================================================
# 10. FILA DE ATENDIMENTO
# ============================================================

class FilaAtendimento:

    def __init__(self):
        self.inicio = None
        self.fim = None
        self.quantidade = 0

    # --------------------------------------------------------
    # Verifica se a fila está vazia
    # --------------------------------------------------------

    def esta_vazia(self):

        return self.inicio is None

    # --------------------------------------------------------
    # Adiciona um paciente
    # --------------------------------------------------------

    def adicionar(self, paciente):

        novo = NodePaciente(paciente)

        # Se a fila estiver vazia
        if self.esta_vazia():

            self.inicio = novo
            self.fim = novo

        # Se o paciente for prioritário
        elif paciente.prioridade == "Prioridade":

            # Se o primeiro paciente também for prioritário,
            # procuramos o final do grupo de prioritários.

            if self.inicio.paciente.prioridade == "Prioridade":

                atual = self.inicio

                while (
                    atual.proximo is not None
                    and atual.proximo.paciente.prioridade == "Prioridade"
                ):
                    atual = atual.proximo

                novo.proximo = atual.proximo
                atual.proximo = novo

                # Se foi inserido no final da fila
                if novo.proximo is None:
                    self.fim = novo

            else:

                # Coloca o paciente prioritário no início
                novo.proximo = self.inicio
                self.inicio = novo

        # Paciente normal
        else:

            self.fim.proximo = novo
            self.fim = novo

        self.quantidade += 1

    # --------------------------------------------------------
    # Atende o primeiro paciente
    # --------------------------------------------------------

    def atender(self):

        if self.esta_vazia():

            print("A fila está vazia.")

            return None

        paciente = self.inicio.paciente

        self.inicio = self.inicio.proximo

        # Se a fila ficou vazia
        if self.inicio is None:
            self.fim = None

        self.quantidade -= 1

        return paciente

    # --------------------------------------------------------
    # Lista os pacientes
    # --------------------------------------------------------

    def listar(self):

        if self.esta_vazia():

            print("A fila está vazia.")

            return

        atual = self.inicio

        contador = 1

        while atual is not None:

            paciente = atual.paciente

            print(
                f"{contador}. "
                f"{paciente.nome} - "
                f"{paciente.idade} anos - "
                f"{paciente.prioridade}"
            )

            atual = atual.proximo
            contador += 1

    # --------------------------------------------------------
    # Retorna o tamanho da fila
    # --------------------------------------------------------

    def tamanho(self):

        return self.quantidade


# ============================================================
# 11. TESTE DO SISTEMA
# ============================================================

print("\n" + "=" * 60)
print("TESTE DO SISTEMA")
print("=" * 60)

fila = FilaAtendimento()

print("\nFila inicialmente:")

fila.listar()

print("\nA fila está vazia?", fila.esta_vazia())

print("Quantidade:", fila.tamanho())


# ============================================================
# 12. ADICIONANDO 5 PACIENTES
# ============================================================

fila.adicionar(Paciente("Ana", 32, "Normal"))
fila.adicionar(Paciente("Bruno", 70, "Normal"))
fila.adicionar(Paciente("Carlos", 45, "Prioridade"))
fila.adicionar(Paciente("Daniel", 25, "Normal"))
fila.adicionar(Paciente("Eduarda", 80, "Prioridade"))


print("\nFila depois de adicionar 5 pacientes:")

fila.listar()

print("\nQuantidade de pacientes:", fila.tamanho())


# ============================================================
# 13. TESTANDO O ATENDIMENTO
# ============================================================

print("\n" + "=" * 60)
print("ATENDIMENTO")
print("=" * 60)

paciente = fila.atender()

if paciente is not None:
    print("Atendendo:", paciente)


paciente = fila.atender()

if paciente is not None:
    print("Atendendo:", paciente)


print("\nFila depois dos atendimentos:")

fila.listar()

print("\nQuantidade:", fila.tamanho())


# ============================================================
# 14. TESTANDO O ATENDIMENTO ATÉ ESVAZIAR
# ============================================================

print("\n" + "=" * 60)
print("ESVAZIANDO A FILA")
print("=" * 60)

while not fila.esta_vazia():

    paciente = fila.atender()

    print("Atendendo:", paciente)


print("\nFila vazia?", fila.esta_vazia())

print("Quantidade:", fila.tamanho())


# ============================================================
# 15. TESTANDO FILA VAZIA
# ============================================================

print("\nTentando atender uma fila vazia:")

paciente = fila.atender()


# ============================================================
# 16. TESTE DO ATENDIMENTO PRIORITÁRIO
# ============================================================

print("\n" + "=" * 60)
print("DESAFIO - ATENDIMENTO PRIORITÁRIO")
print("=" * 60)

fila_prioridade = FilaAtendimento()

fila_prioridade.adicionar(
    Paciente("Ana", 32, "Normal")
)

fila_prioridade.adicionar(
    Paciente("Bruno", 40, "Normal")
)

fila_prioridade.adicionar(
    Paciente("Carlos", 45, "Prioridade")
)

print("\nFila inicial:")

fila_prioridade.listar()

print("\nOrdem de atendimento:")

while not fila_prioridade.esta_vazia():

    paciente = fila_prioridade.atender()

    print(paciente.nome)


# ============================================================
# 17. TESTE COM VÁRIOS PRIORITÁRIOS
# ============================================================

print("\n" + "=" * 60)
print("TESTE COM VÁRIOS PRIORITÁRIOS")
print("=" * 60)

fila_prioridade = FilaAtendimento()

fila_prioridade.adicionar(
    Paciente("Ana", 32, "Normal")
)

fila_prioridade.adicionar(
    Paciente("Bruno", 40, "Prioridade")
)

fila_prioridade.adicionar(
    Paciente("Carlos", 45, "Normal")
)

fila_prioridade.adicionar(
    Paciente("Daniel", 50, "Prioridade")
)

print("\nFila:")

fila_prioridade.listar()

print("\nOrdem esperada:")

print("1. Bruno")
print("2. Daniel")
print("3. Ana")
print("4. Carlos")

print("\nOrdem real:")

while not fila_prioridade.esta_vazia():

    paciente = fila_prioridade.atender()

    print(paciente.nome)


# ============================================================
# 18. TESTANDO A REMOÇÃO DO ÚLTIMO PACIENTE
# ============================================================

print("\n" + "=" * 60)
print("TESTE DA REMOÇÃO DO ÚLTIMO PACIENTE")
print("=" * 60)

fila_teste = FilaAtendimento()

fila_teste.adicionar(
    Paciente("João", 30, "Normal")
)

print("Antes do atendimento:")

fila_teste.listar()

paciente = fila_teste.atender()

print("\nAtendendo:", paciente.nome)

print("\nDepois do atendimento:")

fila_teste.listar()

print("\nFila vazia?", fila_teste.esta_vazia())

print("Quantidade:", fila_teste.tamanho())


# ============================================================
# 19. TESTE DE IDENTIDADE DOS OBJETOS
# ============================================================

print("\n" + "=" * 60)
print("TESTE DE IDENTIDADE")
print("=" * 60)

lista1 = [1, 2, 3]
lista2 = lista1

print("id(lista1):", id(lista1))
print("id(lista2):", id(lista2))

if id(lista1) == id(lista2):
    print("As duas variáveis referenciam o mesmo objeto.")


# ============================================================
# 20. REFLEXÃO FINAL
# ============================================================

print("\n" + "=" * 60)
print("REFLEXÃO FINAL")
print("=" * 60)

print("""
1. O que é uma referência em Python?

Uma referência é a ligação entre um nome e um objeto.
Uma variável em Python pode ser entendida como um nome
associado a um objeto na memória.

Exemplo:

a = [10, 20, 30]
b = a

Nesse caso, a e b referenciam a mesma lista.


2. Qual a diferença entre b = a e b = a.copy()?

Em b = a, as duas variáveis referenciam o mesmo objeto.

Em b = a.copy(), uma nova lista é criada.
Por isso, alterar b não altera a lista original a.


3. Por que uma estrutura encadeada precisa de referências?

Porque cada Node precisa saber qual é o próximo Node.
O atributo proximo guarda uma referência para o próximo
elemento da estrutura.

Exemplo:

A -> B -> C -> None


4. Qual erro foi encontrado durante a depuração?

O erro foi não atualizar a variável atual dentro do while.

Código com erro:

while atual is not None:
    print(atual.valor)

Correção:

while atual is not None:
    print(atual.valor)
    atual = atual.proximo

Sem atualizar atual, o programa fica preso em loop infinito.


5. Como referências ajudam a compreender estruturas dinâmicas?

As referências permitem conectar objetos entre si.
Em uma lista encadeada, cada Node possui uma referência
para outro Node.

Assim, podemos criar estruturas que crescem e diminuem
durante a execução do programa.
""")


# ============================================================
# 21. SÍNTESE DA AULA
# ============================================================

print("\n" + "=" * 60)
print("SÍNTESE")
print("=" * 60)

print("""
VARIÁVEL
   ↓
REFERÊNCIA
   ↓
OBJETO
   ↓
NÓ
   ↓
ESTRUTURA DINÂMICA
   ↓
LISTA / FILA / PILHA
""")


# ============================================================
# 22. CHECKLIST FINAL
# ============================================================

print("=" * 60)
print("CHECKLIST DA ATIVIDADE")
print("=" * 60)

checklist = [
    "Criar Paciente",
    "Criar Node",
    "Criar FilaAtendimento",
    "Implementar adicionar()",
    "Implementar atender()",
    "Implementar listar()",
    "Implementar esta_vazia()",
    "Implementar tamanho()",
    "Testar pelo menos cinco pacientes",
    "Testar fila vazia",
    "Testar um paciente",
    "Testar vários pacientes",
    "Testar atendimento",
    "Testar remoção do último paciente",
    "Testar paciente prioritário",
    "Documentar um erro e sua correção"
]

for item in checklist:
    print("[OK]", item)


print("\n" + "=" * 60)
print("ATIVIDADE CONCLUÍDA")
print("=" * 60)
````

````python
# ============================================================
# AULA 03 - ESTRUTURA DE DADOS
# PONTEIROS, REFERÊNCIAS E ESTRUTURAS DINÂMICAS
# ============================================================


# ============================================================
# EXERCÍCIO 1 - REFERÊNCIAS
# ============================================================

print("=" * 60)
print("EXERCÍCIO 1 - REFERÊNCIAS")
print("=" * 60)

a = [10, 20, 30]
b = a

b.append(40)

print("a:", a)
print("b:", b)
print("id(a):", id(a))
print("id(b):", id(b))
print("a e b são o mesmo objeto?", a is b)

print("\nRespostas:")
print("1. a e b possuem [10, 20, 30, 40].")
print("2. Sim, representam o mesmo objeto.")
print("3. Porque b = a cria uma referência para a mesma lista.")
print("4. Com b = a.copy(), seria criada uma nova lista.")


# ============================================================
# REFERÊNCIA X CÓPIA
# ============================================================

print("\n" + "=" * 60)
print("REFERÊNCIA X CÓPIA")
print("=" * 60)

a = [10, 20, 30]
b = a

b.append(40)

print("Usando b = a:")
print("a:", a)
print("b:", b)
print("Mesmo objeto?", a is b)

a = [10, 20, 30]
b = a.copy()

b.append(40)

print("\nUsando b = a.copy():")
print("a:", a)
print("b:", b)
print("Mesmo objeto?", a is b)


# ============================================================
# MUTABILIDADE
# ============================================================

print("\n" + "=" * 60)
print("MUTABILIDADE")
print("=" * 60)

lista = [1, 2, 3]

print("Antes:", lista)

lista.append(4)

print("Depois:", lista)

print("\nObjetos mutáveis:")
print("- list")
print("- dict")
print("- set")

print("\nObjetos imutáveis:")
print("- int")
print("- float")
print("- str")
print("- tuple")
print("- bool")


# ============================================================
# PASSAGEM DE OBJETOS PARA FUNÇÕES
# ============================================================

print("\n" + "=" * 60)
print("PASSAGEM DE OBJETOS PARA FUNÇÕES")
print("=" * 60)


def adicionar_item(lista):
    lista.append(100)


numeros = [1, 2, 3]

print("Antes:", numeros)

adicionar_item(numeros)

print("Depois:", numeros)


# ============================================================
# CLASSE NODE
# ============================================================

class Node:

    def __init__(self, valor):
        self.valor = valor
        self.proximo = None


# ============================================================
# EXERCÍCIO 2 - CONSTRUINDO UMA CADEIA
# ============================================================

print("\n" + "=" * 60)
print("EXERCÍCIO 2 - CONSTRUINDO UMA CADEIA")
print("=" * 60)

n1 = Node("A")
n2 = Node("B")
n3 = Node("C")

n1.proximo = n2
n2.proximo = n3

print("Estrutura:")
print("A -> B -> C -> None")

print("\nPercorrendo:")

atual = n1

while atual is not None:
    print(atual.valor)
    atual = atual.proximo


# ============================================================
# EXERCÍCIO 3 - DEPURAÇÃO
# ============================================================

print("\n" + "=" * 60)
print("EXERCÍCIO 3 - DEPURAÇÃO")
print("=" * 60)

n1 = Node(10)
n2 = Node(20)
n3 = Node(30)

n1.proximo = n2
n2.proximo = n3

atual = n1

while atual is not None:
    print(atual.valor)
    atual = atual.proximo

print("\nProblema:")
print("A variável atual não era atualizada.")

print("\nCorreção:")
print("atual = atual.proximo")

print("\nSaída correta:")
print("10")
print("20")
print("30")


# ============================================================
# LISTA ENCADEADA
# ============================================================

print("\n" + "=" * 60)
print("LISTA ENCADEADA")
print("=" * 60)


class ListaEncadeada:

    def __init__(self):
        self.inicio = None

    def adicionar(self, valor):
        novo = Node(valor)
        novo.proximo = self.inicio
        self.inicio = novo

    def exibir(self):
        atual = self.inicio

        while atual is not None:
            print(atual.valor)
            atual = atual.proximo


lista = ListaEncadeada()

lista.adicionar(10)
lista.adicionar(20)
lista.adicionar(30)

print("Elementos da lista:")

lista.exibir()


# ============================================================
# CLASSE PACIENTE
# ============================================================

print("\n" + "=" * 60)
print("SISTEMA DE ATENDIMENTO DA CLÍNICA")
print("=" * 60)


class Paciente:

    def __init__(self, nome, idade, prioridade):
        self.nome = nome
        self.idade = idade
        self.prioridade = prioridade

    def __str__(self):
        return (
            self.nome
            + " - "
            + str(self.idade)
            + " anos - "
            + self.prioridade
        )


# ============================================================
# NODE DA FILA
# ============================================================

class NodePaciente:

    def __init__(self, paciente):
        self.paciente = paciente
        self.proximo = None


# ============================================================
# FILA DE ATENDIMENTO
# ============================================================

class FilaAtendimento:

    def __init__(self):
        self.inicio = None
        self.fim = None
        self.quantidade = 0

    def esta_vazia(self):

        return self.inicio is None

    def adicionar(self, paciente):

        novo = NodePaciente(paciente)

        # Fila vazia
        if self.esta_vazia():

            self.inicio = novo
            self.fim = novo

        # Paciente prioritário
        elif paciente.prioridade == "Prioridade":

            # Se o primeiro não for prioritário,
            # coloca o novo no início.
            if self.inicio.paciente.prioridade != "Prioridade":

                novo.proximo = self.inicio
                self.inicio = novo

            else:

                # Procura o último paciente prioritário
                atual = self.inicio

                while (
                    atual.proximo is not None
                    and atual.proximo.paciente.prioridade == "Prioridade"
                ):
                    atual = atual.proximo

                novo.proximo = atual.proximo
                atual.proximo = novo

                if novo.proximo is None:
                    self.fim = novo

        # Paciente normal
        else:

            self.fim.proximo = novo
            self.fim = novo

        self.quantidade += 1

    def atender(self):

        if self.esta_vazia():

            print("A fila está vazia.")

            return None

        paciente = self.inicio.paciente

        self.inicio = self.inicio.proximo

        if self.inicio is None:
            self.fim = None

        self.quantidade -= 1

        return paciente

    def listar(self):

        if self.esta_vazia():

            print("A fila está vazia.")

            return

        atual = self.inicio
        numero = 1

        while atual is not None:

            paciente = atual.paciente

            print(
                str(numero)
                + ". "
                + paciente.nome
                + " - "
                + str(paciente.idade)
                + " anos - "
                + paciente.prioridade
            )

            atual = atual.proximo
            numero += 1

    def tamanho(self):

        return self.quantidade


# ============================================================
# TESTE COM 5 PACIENTES
# ============================================================

print("\n" + "=" * 60)
print("TESTE COM 5 PACIENTES")
print("=" * 60)

fila = FilaAtendimento()

fila.adicionar(Paciente("Ana", 32, "Normal"))
fila.adicionar(Paciente("Bruno", 70, "Normal"))
fila.adicionar(Paciente("Carlos", 45, "Prioridade"))
fila.adicionar(Paciente("Daniel", 25, "Normal"))
fila.adicionar(Paciente("Eduarda", 80, "Prioridade"))

print("\nFila:")

fila.listar()

print("\nQuantidade:", fila.tamanho())


# ============================================================
# ATENDIMENTO
# ============================================================

print("\n" + "=" * 60)
print("ATENDIMENTO")
print("=" * 60)

paciente = fila.atender()

if paciente is not None:
    print("Atendendo:", paciente)


paciente = fila.atender()

if paciente is not None:
    print("Atendendo:", paciente)


print("\nFila depois dos atendimentos:")

fila.listar()

print("\nQuantidade:", fila.tamanho())


# ============================================================
# TESTE DE FILA VAZIA
# ============================================================

print("\n" + "=" * 60)
print("TESTE DE FILA VAZIA")
print("=" * 60)

while not fila.esta_vazia():

    paciente = fila.atender()

    print("Atendendo:", paciente)


print("\nA fila está vazia?", fila.esta_vazia())
print("Quantidade:", fila.tamanho())

print("\nTentando atender novamente:")

fila.atender()


# ============================================================
# DESAFIO - ATENDIMENTO PRIORITÁRIO
# ============================================================

print("\n" + "=" * 60)
print("DESAFIO - ATENDIMENTO PRIORITÁRIO")
print("=" * 60)

fila = FilaAtendimento()

fila.adicionar(Paciente("Ana", 32, "Normal"))
fila.adicionar(Paciente("Bruno", 40, "Normal"))
fila.adicionar(Paciente("Carlos", 45, "Prioridade"))

print("\nFila:")

fila.listar()

print("\nOrdem de atendimento:")

while not fila.esta_vazia():

    paciente = fila.atender()

    print(paciente.nome)


# ============================================================
# TESTE COM VÁRIOS PRIORITÁRIOS
# ============================================================

print("\n" + "=" * 60)
print("TESTE COM VÁRIOS PRIORITÁRIOS")
print("=" * 60)

fila = FilaAtendimento()

fila.adicionar(Paciente("Ana", 32, "Normal"))
fila.adicionar(Paciente("Bruno", 40, "Prioridade"))
fila.adicionar(Paciente("Carlos", 45, "Normal"))
fila.adicionar(Paciente("Daniel", 50, "Prioridade"))

print("\nFila:")

fila.listar()

print("\nOrdem esperada:")
print("Bruno")
print("Daniel")
print("Ana")
print("Carlos")

print("\nOrdem de atendimento:")

while not fila.esta_vazia():

    paciente = fila.atender()

    print(paciente.nome)


# ============================================================
# TESTE DE REMOÇÃO DO ÚLTIMO PACIENTE
# ============================================================

print("\n" + "=" * 60)
print("TESTE DE REMOÇÃO DO ÚLTIMO PACIENTE")
print("=" * 60)

fila = FilaAtendimento()

fila.adicionar(Paciente("João", 30, "Normal"))

print("\nAntes:")

fila.listar()

paciente = fila.atender()

print("\nAtendendo:", paciente.nome)

print("\nDepois:")

fila.listar()

print("\nFila vazia?", fila.esta_vazia())
print("Quantidade:", fila.tamanho())


# ============================================================
# REFLEXÃO FINAL
# ============================================================

print("\n" + "=" * 60)
print("REFLEXÃO FINAL")
print("=" * 60)

print("""
1. O que é uma referência em Python?

Uma referência é uma ligação entre um nome e um objeto.
Quando duas variáveis referenciam o mesmo objeto, uma alteração
nesse objeto pode ser observada pelas duas variáveis.

2. Qual a diferença entre b = a e b = a.copy()?

b = a faz as duas variáveis referenciarem o mesmo objeto.

b = a.copy() cria uma nova lista, independente da original.

3. Por que uma estrutura encadeada precisa de referências?

Porque cada nó precisa guardar uma referência para o próximo nó.
Assim os nós podem ser conectados formando uma estrutura.

4. Qual erro foi encontrado durante a depuração?

O erro foi não atualizar a variável atual dentro do while.

A correção foi adicionar:

atual = atual.proximo

5. Como referências ajudam a compreender estruturas dinâmicas?

As referências permitem conectar diferentes objetos.
Em uma lista encadeada, um Node referencia o próximo Node.
Isso permite criar estruturas que podem crescer e diminuir
durante a execução do programa.
""")


# ============================================================
# CHECKLIST
# ============================================================

print("=" * 60)
print("CHECKLIST")
print("=" * 60)

print("[OK] Criar Paciente")
print("[OK] Criar Node")
print("[OK] Criar FilaAtendimento")
print("[OK] Implementar adicionar()")
print("[OK] Implementar atender()")
print("[OK] Implementar listar()")
print("[OK] Implementar esta_vazia()")
print("[OK] Implementar tamanho()")
print("[OK] Testar cinco pacientes")
print("[OK] Testar fila vazia")
print("[OK] Testar atendimento")
print("[OK] Testar remoção do último paciente")
print("[OK] Testar paciente prioritário")
print("[OK] Documentar erro e correção")


print("\n" + "=" * 60)
print("ATIVIDADE CONCLUÍDA!")
print("=" * 60)
````
